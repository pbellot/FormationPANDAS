```python
import pandas as pd
import numpy as np
```

```python
# Charger le fichier CSV dans un DataFrame
df = pd.read_csv('Datasets/titanic.csv')

# Afficher les premières lignes du DataFrame
df.head()
```

```python
# Intervertir les positions des colonnes Sex et Age sans modifier les valeurs
cols = df.columns.tolist()
# Trouver les indices de Sex et Age
idx_sex = cols.index('Sex')
idx_age = cols.index('Age')
# Intervertir les positions dans la liste
cols[idx_sex], cols[idx_age] = cols[idx_age], cols[idx_sex]
# Réorganiser le DataFrame
df = df[cols]

# Afficher les premières lignes après l'interversion
df.head()
```

```python
# Obtenir les valeurs uniques d'âge avec la méthode unique() de Pandas
ages_uniques = df['Age'].unique()

# Ordonner les valeurs avec la méthode sort() de NumPy
ages_tries = np.sort(ages_uniques)

print("Valeurs d'âge triées:")
print(ages_tries)
```

```python
# Utiliser sort_values() pour trier par Age
df_sorted = df.sort_values('Age')

# Les 3 passagers les plus jeunes
print("3 passagers les plus jeunes (avec sort_values() et head()):")
df_sorted.head(3)
```

```python
# Les 3 passagers les plus âgés
df_sorted.tail(3)
```

```python
# Utiliser query() pour trouver les passagers de 3ème classe qui ont survécu
result = df.query('Pclass == 3 and Survived == 1')

print(f"Nombre de passagers de 3ème classe qui ont survécu: {len(result)}")
print("\nDétails:")
result
```

```python
# Prix moyen pour les passagers de 3ème classe qui ont survécu
fare_survived = df.query('Pclass == 3 and Survived == 1')['Fare'].mean()

# Prix moyen pour les passagers de 3ème classe qui n'ont pas survécu
fare_not_survived = df.query('Pclass == 3 and Survived == 0')['Fare'].mean()

print("Prix moyen du billet pour les passagers de 3ème classe:")
print(f"  - Qui ont survécu: {fare_survived:.2f}")
print(f"  - Qui n'ont pas survécu: {fare_not_survived:.2f}")
print(f"\nDifférence: {fare_survived - fare_not_survived:.2f}")
```

```python
# En utilisant groupby()
# Filtrer les passagers de 3ème classe
third_class = df.query('Pclass == 3')

# Grouper par Survived et calculer la moyenne de Fare
fare_by_survival = third_class.groupby('Survived')['Fare'].mean()

print("Pour les passagers de 3ème classe (avec groupby()):")
print(f"  - Qui n'ont pas survécu (0): {fare_by_survival[0]:.2f}")
print(f"  - Qui ont survécu (1): {fare_by_survival[1]:.2f}")
```

```python
# Expérimenter la méthode get_dummies() pour transformer les variables catégorielles
print("Avant get_dummies():")
print(f"Colonnes du dataframe: {df.columns.tolist()}")
print(f"Types de données:\n{df.dtypes}")

print("\n" + "="*60 + "\n")

# Appliquer get_dummies() sur les colonnes catégorielles
df_dummies = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=False)

print("Après get_dummies():")
print(f"Colonnes du dataframe: {df_dummies.columns.tolist()}")
print(f"\nNouveaux types de données:\n{df_dummies.dtypes}")

print("\n" + "="*60 + "\n")

# Afficher les premières lignes
print("Premières lignes après transformation:")
df_dummies[['PassengerId', 'Sex_female', 'Sex_male', 'Embarked_C', 'Embarked_Q', 'Embarked_S']].head(10)
```

```python
# Appliquer K-means sur les données du Titanic (sans utiliser Survived)
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, accuracy_score, homogeneity_score

print("="*60)
print("CLUSTERING K-MEANS SUR LES DONNÉES DU TITANIC")
print("="*60)

# Préparer les données: sélectionner les variables pertinentes
# On utilise les variables numériques et les dummies des variables catégorielles
features_to_use = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'Sex_female', 'Sex_male', 
                   'Embarked_C', 'Embarked_Q', 'Embarked_S']

# Créer un dataframe avec uniquement les features et supprimer les NaN
df_kmeans = df_dummies[features_to_use].copy()
df_kmeans = df_kmeans.dropna()

print(f"\nNombre d'observations utilisées: {len(df_kmeans)}")
print(f"Nombre de variables: {len(features_to_use)}")
print(f"Variables utilisées: {features_to_use}\n")

# Normaliser les données (important pour K-means)
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_kmeans)

print("Données normalisées (premiers 5 lignes):")
print(df_scaled[:5])

print("\n" + "="*60 + "\n")

# Appliquer K-means avec k=2
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
clusters = kmeans.fit_predict(df_scaled)

print("RÉSULTATS K-MEANS (k=2):")
print(f"Inertie (somme des distances): {kmeans.inertia_:.2f}")
print(f"Distribution des clusters: {np.bincount(clusters)}")

print("\n" + "="*60 + "\n")

# Récupérer les vraies valeurs de Survived pour les mêmes observations
true_survived = df_dummies.loc[df_kmeans.index, 'Survived'].values

# Créer une matrice de confusion
print("Matrice de confusion (clusters vs vraie classe Survived):")
cm = confusion_matrix(true_survived, clusters)
print(cm)

# Calculer l'accuracy
accuracy = accuracy_score(true_survived, clusters)
print(f"\nAccuracy brute: {accuracy:.4f}")

# Essayer aussi avec les clusters inversés pour voir le meilleur mapping
accuracy_inverted = accuracy_score(true_survived, 1 - clusters)
print(f"Accuracy avec clusters inversés: {accuracy_inverted:.4f}")

# Utiliser l'accuracy la meilleure
best_accuracy = max(accuracy, accuracy_inverted)
print(f"\nMeilleure accuracy: {best_accuracy:.4f}")

# Calculer l'homogénéité (mesure de la qualité du clustering)
homogeneity = homogeneity_score(true_survived, clusters)
print(f"Score d'homogénéité: {homogeneity:.4f}")

print("\n" + "="*60 + "\n")

# Visualiser les résultats avec un scatter plot (sur les 2 premières composantes principales)
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_scaled)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Clusters trouvés par K-means
axes[0].scatter(df_pca[:, 0], df_pca[:, 1], c=clusters, cmap='viridis', alpha=0.6, s=50)
centroides_pca = pca.transform(kmeans.cluster_centers_)
axes[0].scatter(centroides_pca[:, 0], centroides_pca[:, 1], 
                marker='X', s=500, c='red', edgecolors='black', linewidths=2)
axes[0].set_title('Clusters trouvés par K-means')
axes[0].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%})')
axes[0].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%})')

# Plot 2: Vraie classe Survived
axes[1].scatter(df_pca[:, 0], df_pca[:, 1], c=true_survived, cmap='RdYlGn', alpha=0.6, s=50)
axes[1].set_title('Vraie classe Survived (0=Non survécu, 1=Survécu)')
axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%})')
axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%})')

plt.tight_layout()
plt.show()

print(f"Variance expliquée par PC1 et PC2: {(pca.explained_variance_ratio_[0] + pca.explained_variance_ratio_[1]):.2%}")
```

```python
# Analyse en Composantes Principales (PCA)
from sklearn.decomposition import PCA

print("="*60)
print("ANALYSE EN COMPOSANTES PRINCIPALES (PCA)")
print("="*60)

# Appliquer PCA avec toutes les composantes
pca_full = PCA()
pca_full.fit(df_scaled)

# Afficher la variance expliquée par chaque composante
print("\nVariance expliquée par chaque composante principale:")
print(f"{'Composante':<15} {'Variance':<15} {'% Variance':<15} {'% Cumulatif':<15}")
print("-" * 60)

cumsum_variance = 0
for i, var in enumerate(pca_full.explained_variance_ratio_):
    cumsum_variance += var
    print(f"PC{i+1:<13} {var:<15.6f} {var*100:<14.2f}% {cumsum_variance*100:<14.2f}%")

print("\n" + "="*60 + "\n")

# Déterminer le nombre de composantes pour expliquer 95% de la variance
n_components_95 = np.argmax(np.cumsum(pca_full.explained_variance_ratio_) >= 0.95) + 1
print(f"Nombre de composantes pour expliquer 95% de variance: {n_components_95}")
print(f"Nombre de composantes pour expliquer 90% de variance: {np.argmax(np.cumsum(pca_full.explained_variance_ratio_) >= 0.90) + 1}")
print(f"Nombre de composantes pour expliquer 80% de variance: {np.argmax(np.cumsum(pca_full.explained_variance_ratio_) >= 0.80) + 1}")

print("\n" + "="*60 + "\n")

# Visualiser la variance expliquée
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Graphique 1: Variance expliquée par composante
axes[0].bar(range(1, len(pca_full.explained_variance_ratio_) + 1), 
            pca_full.explained_variance_ratio_ * 100, alpha=0.7, color='steelblue')
axes[0].set_xlabel('Composante Principale')
axes[0].set_ylabel('Variance Expliquée (%)')
axes[0].set_title('Variance expliquée par chaque composante')
axes[0].set_xticks(range(1, len(pca_full.explained_variance_ratio_) + 1))

# Graphique 2: Variance cumulée
cumsum_var = np.cumsum(pca_full.explained_variance_ratio_ * 100)
axes[1].plot(range(1, len(pca_full.explained_variance_ratio_) + 1), cumsum_var, 
             marker='o', linestyle='-', color='darkblue', linewidth=2)
axes[1].axhline(y=95, color='r', linestyle='--', label='95% variance')
axes[1].axhline(y=90, color='orange', linestyle='--', label='90% variance')
axes[1].axhline(y=80, color='green', linestyle='--', label='80% variance')
axes[1].set_xlabel('Nombre de Composantes')
axes[1].set_ylabel('Variance Cumulée (%)')
axes[1].set_title('Variance cumulée')
axes[1].set_xticks(range(1, len(pca_full.explained_variance_ratio_) + 1))
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n" + "="*60 + "\n")

# Appliquer PCA avec un nombre réduit de composantes (ex: 3)
pca_3 = PCA(n_components=3)
data_pca_3 = pca_3.fit_transform(df_scaled)

print("Matrice des loadings (composante 1 et 2):")
loadings = pd.DataFrame(
    pca_3.components_[:2].T,
    columns=['PC1', 'PC2'],
    index=features_to_use
)
print(loadings)
print(f"\nVariance expliquée par les 3 premières composantes: {pca_3.explained_variance_ratio_.sum():.2%}")

print("\n" + "="*60 + "\n")

# Visualiser les données en 3D (PC1, PC2, PC3)
print("Visualisation 2D (PC1 vs PC2):")
plt.figure(figsize=(10, 7))
scatter = plt.scatter(data_pca_3[:, 0], data_pca_3[:, 1], c=true_survived, 
                     cmap='RdYlGn', alpha=0.6, s=50, edgecolors='black', linewidth=0.5)
plt.xlabel(f'PC1 ({pca_3.explained_variance_ratio_[0]:.2%})')
plt.ylabel(f'PC2 ({pca_3.explained_variance_ratio_[1]:.2%})')
plt.title('Données projetées sur les 2 premières composantes principales')
plt.colorbar(scatter, label='Survived (0=Non, 1=Oui)')

# Ajouter les vecteurs des variables originales
for i, feature in enumerate(features_to_use):
    plt.arrow(0, 0, pca_3.components_[0, i]*3, pca_3.components_[1, i]*3,
              head_width=0.1, head_length=0.1, fc='darkgray', ec='darkgray', alpha=0.5)
    plt.text(pca_3.components_[0, i]*3.2, pca_3.components_[1, i]*3.2, feature, 
             fontsize=9, ha='center', va='center')

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nBiplot montrant l'influence des variables originales sur les composantes")
print("(Les flèches grises montrent la direction et l'intensité de chaque variable)")
```
